```python
def dist(p1, p2):
    return ((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2) ** 0.5

def dbscan_small(p, eps, min_sample):
    names = list(p)

    # Step 1: neighbourhoods (what your code already does)
    nb = {i: {j for j in names if dist(p[i], p[j]) <= eps} for i in names}

    # Step 2: core points = at least min_sample neighbours (counting itself)
    core = {i for i in names if len(nb[i]) >= min_sample}

    # Step 3: grow clusters from unvisited core points
    label, cid = {}, 0
    for i in names:
        if i in label or i not in core:
            continue
        cid += 1
        label[i] = cid
        stack = [i]
        while stack:
            q = stack.pop()
            for r in nb[q]:
                if r not in label:
                    label[r] = cid
                    if r in core:          # only core points keep expanding
                        stack.append(r)

    # Step 4: anything unlabelled is noise (0)
    for i in names:
        label.setdefault(i, 0)
    return nb, core, label


p = {"p1":(1,1), "p2":(1,2), "p3":(2,2), "p4":(2,1), "p5":(3.4,2), "p6":(8,8)}
nb, core, label = dbscan_small(p, eps=1.5, min_sample=3)

print("point  neighbours            type     cluster")
for i in p:
    t = "core" if i in core else ("border" if label[i] else "noise")
    print(f"{i:5}  {', '.join(sorted(nb[i])):20}  {t:7}  {label[i]}")
```
