```python
def dist(p1,p2):
    return round((((p1[0]-p2[0])**2+((p1[1]-p2[1])**2))**(1/2)),2)
```

```python
p={"p1":(1,1),"p2":(1,2),"p3":(2,2),"p4":(2,1),"p5":(3.4,2),"p6":(8,8)}
clust=dict()
eps=1.5
min_sample=3

print("p",end='\t')
for i in p.keys():
    print(i,end="\t")
print("neighbours")

for i in p.keys():
    print(i,end="\t")
    clust[i]=set()
    for j in p.keys():
        d=dist(p[i],p[j])
        print(d,end="\t")
        if eps>d:
            clust[i].add(j)
    [print(k,end=", ") for k in clust[i]]
    print()
```
