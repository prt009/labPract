### Create the `Sales` DataFrame

```python
import pandas as pd

Sales = pd.DataFrame(
    {
        2014: [100.5, 150.8, 200.9, 30000, 40000],
        2015: [12000, 18000, 22000, 30000, 45000],
        2016: [20000, 50000, 70000, 100000, 125000],
        2017: [50000, 60000, 70000, 80000, 90000]
    },
    index=["Madhu", "Kusum", "Kinshuk", "Ankit", "Shruti"]
)

Sales
```

```python
print("Dimensions:", Sales.ndim)
print("Shape:", Sales.shape)
print("Size:", Sales.size)
print("Values:")
print(Sales.values)
```

### Q6 — Create the 2018 sales dictionary

```python
sales_dict = {
    "Madhu": 160000,
    "Kusum": 110000,
    "Kinshuk": 500000,
    "Ankit": 340000,
    "Shruti": 900000
}

sales_dict
```

### Q7 — Create `Sales2` from the dictionary

```python
Sales2 = pd.DataFrame.from_dict(
    sales_dict,
    orient="index",
    columns=[2018]
)

Sales2
```

### Q8 — Check whether `Sales2` is empty and append it to `Sales`

```python
print("Is Sales2 empty?", Sales2.empty)

if not Sales2.empty:
    Sales = pd.concat([Sales, Sales2], axis=1)

Sales
```

### Q9 — Transpose `Sales`

```python
Sales = Sales.T
Sales
```

### Q10 — Display sales made by all salespersons in 2017

```python
Sales.loc[2017]
```

### Q11 — Display selected sales

##### Madhu and Ankit in 2017 and 2018

```python
print("Madhu and Ankit in 2017 and 2018:")
print(Sales.loc[[2017, 2018], ["Madhu", "Ankit"]])

print("\nShruti in 2016:")
print(Sales.loc[2016, "Shruti"])
```

### Q14 — Delete salesperson Kinshuk

### Q15 — Rename Ankit to Vivaan and Madhu to Shailesh

```python
Sales = Sales.rename(
    columns={
        "Ankit": "Vivaan",
        "Madhu": "Shailesh"
    }
)

Sales
```

### Q16 — Update Shailesh's 2018 sales to 100000

```python
Sales.loc[2018, "Shailesh"] = 100000
Sales
```

### Q1.17 & Q1.18 — Write `Sales` to `SalesFigures.txt`
