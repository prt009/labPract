```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
X = np.array([2, 4, 6, 8, 10]).reshape(-1, 1)
Y = np.array([21, 27, 29, 64, 86])

model = LinearRegression()
model.fit(X, Y)

slope = model.coef_[0]
intercept = model.intercept_

print(f"Regression Coefficients:")
print(f"Slope (m): {slope:.2f}")
print(f"Intercept (c): {intercept:.2f}")
print(f"Equation: Y = {slope:.2f}X + ({intercept:.2f})\n")

X_predict = np.array([[12]])
Y_predict = model.predict(X_predict)[0]
print(f"Predicted value of Y for X=12: {Y_predict:.2f}")

plt.scatter(X, Y, color='red', label='Data points')
plt.plot(X, model.predict(X), color='blue', label='Regression line')
plt.scatter(12, Y_predict, color='green', marker='x', s=100, label=f'Prediction at X=12 ({Y_predict:.1f})')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Least Squares Linear Regression')
plt.legend()
plt.grid(True)
plt.show()
```
