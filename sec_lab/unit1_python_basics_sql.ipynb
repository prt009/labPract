## Python Basics and Functions


```python
nm = "Meera"
ag = 19
pct = 72.5
stu = True

print("Name:", nm)
print("Age:", ag)
print("Percentage:", pct)
print("Is student:", stu)

print(type(nm))
print(type(ag))
print(type(pct))
print(type(stu))
```

```python
x = 17
y = 5
print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Floor division:", x // y)
print("Remainder:", x % y)
print("Power:", x ** y)
```

```python
sb = ["Java", "Maths", "Stats"]
print("Subjects:", sb)
print("First:", sb[0])
print("Last:", sb[-1])

sb.append("English")
print("After adding:", sb)
print("Count:", len(sb))

mk = {"Java": 67, "Maths": 91, "Stats": 58}
print("Marks:", mk)
print("Maths marks:", mk["Maths"])

mk["English"] = 84
print("After adding English:", mk)
```

```python
sc = 62
if sc >= 80:
    print("Grade A")
elif sc >= 60:
    print("Grade B")
else:
    print("Grade C")

for s in sb:
    print(s)

for i in range(1, 6):
    print(i)

c = 3
while c > 0:
    print("Countdown:", c)
    c = c - 1
print("Done!")
```

```python
def hello(p):
    print("Hello", p)

hello("Meera")

def add(a, b):
    return a + b

print("8 + 6 =", add(8, 6))

def grade(sc):
    if sc >= 80:
        return "A"
    elif sc >= 60:
        return "B"
    else:
        return "C"

for s in mk:
    print(s, "->", mk[s], "-> Grade", grade(mk[s]))
```

## Python and SQL Database


```python
import sqlite3

con = sqlite3.connect("lib.db")
cur = con.cursor()
print("Database connected!")

cur.execute("DROP TABLE IF EXISTS books")
cur.execute("CREATE TABLE books (id INTEGER PRIMARY KEY, title TEXT, genre TEXT, price INTEGER)")
print("Table created!")
```

```python
cur.execute("INSERT INTO books (title, genre, price) VALUES ('Dune', 'SciFi', 350)")
cur.execute("INSERT INTO books (title, genre, price) VALUES ('Emma', 'Classic', 220)")
cur.execute("INSERT INTO books (title, genre, price) VALUES ('Heat', 'Thriller', 410)")
cur.execute("INSERT INTO books (title, genre, price) VALUES ('Gone', 'Thriller', 300)")
cur.execute("INSERT INTO books (title, genre, price) VALUES ('Ulysses', 'Classic', 180)")
cur.execute("INSERT INTO books (title, genre, price) VALUES ('Cosmos', 'Science', 275)")
con.commit()
print("6 rows inserted!")

cur.execute("SELECT * FROM books")
print("All books:")
for r in cur.fetchall():
    print(r)
```

```python
cur.execute("SELECT title, price FROM books WHERE price > 250")
print("Books above 250:")
for r in cur.fetchall():
    print(r)

cur.execute("SELECT title, price FROM books ORDER BY price DESC")
print("Highest to lowest price:")
for r in cur.fetchall():
    print(r)
```

```python
cur.execute("UPDATE books SET price = 250 WHERE title = 'Emma'")
con.commit()
print("Emma price updated to 250")

cur.execute("DELETE FROM books WHERE title = 'Ulysses'")
con.commit()
print("Ulysses deleted")

cur.execute("SELECT COUNT(*) FROM books")
print("Total books:", cur.fetchone()[0])

cur.execute("SELECT AVG(price) FROM books")
print("Average price:", cur.fetchone()[0])

cur.execute("SELECT * FROM books")
print("Final table:")
for r in cur.fetchall():
    print(r)

con.close()
print("Connection closed.")
```
