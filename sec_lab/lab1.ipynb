```python
conn = sql.connect("data.db")
crsr = conn.cursor()
print("database connected")
```

```python
crsr.execute("create table if not exists students (id int, name varchar(20), age int);")
print("'students' table created")
```

```python
id,name,age=3,"ppt",19
crsr.execute(f"insert into students (id,name,age) values({id},'{name}',{age});")
print("1 row inserted")
conn.commit()
```

```python
crsr.execute("select * from students;")
students_table=crsr.fetchall()
for i in students_table:
    print(i)
```
